```python
# ╔══════════════════════════════════════════════════════════════════════════════╗
# ║                        ║
# ║   "Optimized CNN Models for Skin Lesion Classification"                     ║
#         ║
# ║   Experiment 4: Balanced Dataset + Tuned Transfer Learning Model            ║
# ║   Dataset  : Skin Cancer MNIST — HAM10000 (7 classes, 10,015 images)        ║
# ║   Split    : 80 % Train / 10 % Val / 10 % Test                             ║
# ║   Optimizer: Adamax  |  LR: 0.001 with LR Scheduler  |  Epochs: 30         ║
# ║   Architecture: ResNet50 + Dense[512→128→64→32] + BN + Softmax           ║
# ╚══════════════════════════════════════════════════════════════════════════════╝
#
# in the pipeline has been changed.
#
# ── IMPORTS ───────────────────────────────────────────────────────────────────
import os, warnings, zipfile
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches
import seaborn as sns
from itertools import cycle

from sklearn.model_selection import train_test_split
from sklearn.metrics import (classification_report, confusion_matrix,
                             roc_curve, auc)
from sklearn.preprocessing import label_binarize

import tensorflow as tf
from tensorflow.keras.applications.resnet50 import ResNet50, preprocess_input as resnet_preprocess
from tensorflow.keras.models import Model
from tensorflow.keras.layers import (Dense, Dropout, GlobalAveragePooling2D,
                                     BatchNormalization, Activation)
from tensorflow.keras.optimizers import Adamax
from tensorflow.keras.callbacks import (LearningRateScheduler,
                                        ModelCheckpoint, EarlyStopping)
from tensorflow.keras.preprocessing.image import ImageDataGenerator

warnings.filterwarnings('ignore')
tf.get_logger().setLevel('ERROR')

print("=" * 65)
print("  ResNet50 — Paper 1 Full Pipeline Replication")
print("=" * 65)
print(f"  TensorFlow : {tf.__version__}")
print(f"  GPU(s)     : {tf.config.list_physical_devices('GPU')}")
print("=" * 65)


# ══════════════════════════════════════════════════════════════════════════════
# SECTION 1 — CONFIGURATION  (Paper 1 §2.2, §2.3, §2.4)
# ══════════════════════════════════════════════════════════════════════════════
IMG_SIZE         = 224          # §2.2: "image input size was transformed to 224×224×3"
BATCH_SIZE       = 32
EPOCHS           = 40          # §2.4: "model was trained for … 30 epochs"
INITIAL_LR       = 1e-3         # §2.4: "initial rate of 0.001"
NUM_CLASSES      = 7
RANDOM_SEED      = 42
TARGET_PER_CLASS = 6100         # §2.2: augmented to equalise all classes (~6000+)

CLASS_NAMES = ['akiec', 'bcc', 'bkl', 'df', 'mel', 'nv', 'vasc']
CLASS_TO_IDX = {c: i for i, c in enumerate(CLASS_NAMES)}

# ── Kaggle HAM10000 paths ─────────────────────────────────────────────────────
BASE_DIR  = '/kaggle/input/datasets/kmader/skin-cancer-mnist-ham10000'
IMG_DIR1  = os.path.join(BASE_DIR, 'HAM10000_images_part_1')
IMG_DIR2  = os.path.join(BASE_DIR, 'HAM10000_images_part_2')
META_PATH = os.path.join(BASE_DIR, 'HAM10000_metadata.csv')
OUT_DIR   = '/kaggle/working'

# ── Directory for all saved figures ──────────────────────────────────────────
FIG_DIR = os.path.join(OUT_DIR, 'thesis_figures')
os.makedirs(FIG_DIR, exist_ok=True)

def save_fig(fname):
    """Save current figure to FIG_DIR and print confirmation."""
    path = os.path.join(FIG_DIR, fname)
    plt.savefig(path, dpi=150, bbox_inches='tight')
    plt.show()
    print(f"  Saved → {fname}")
    return path


# ══════════════════════════════════════════════════════════════════════════════
# SECTION 2 — LOAD & MAP METADATA
# ══════════════════════════════════════════════════════════════════════════════
print("\n[1/9] Loading HAM10000 metadata …")

df = pd.read_csv(META_PATH)

def find_path(img_id):
    for d in [IMG_DIR1, IMG_DIR2]:
        p = os.path.join(d, img_id + '.jpg')
        if os.path.exists(p):
            return p
    return None

df['path']  = df['image_id'].apply(find_path)
df = df[df['path'].notna()].reset_index(drop=True)
df['label'] = df['dx'].map(CLASS_TO_IDX)

print(f"  Images located : {len(df):,}")
print("  Raw class distribution:")
for cls in CLASS_NAMES:
    n = (df['dx'] == cls).sum()
    print(f"    {cls:>6} : {n:>5}")

# ── Graph 1 — Raw Class Distribution Bar Chart ────────────────────────────────
raw_counts = [( df['dx'] == cls).sum() for cls in CLASS_NAMES]
colors_bar = ['#4C72B0','#DD8452','#55A868','#C44E52','#8172B2','#937860','#DA8BC3']

fig, ax = plt.subplots(figsize=(9, 5))
bars = ax.bar(CLASS_NAMES, raw_counts, color=colors_bar, edgecolor='white', linewidth=0.8)
for bar, val in zip(bars, raw_counts):
    ax.text(bar.get_x() + bar.get_width()/2, bar.get_height() + 40,
            str(val), ha='center', va='bottom', fontsize=10, fontweight='bold')
ax.set_title('ResNet50 — Raw Class Distribution', fontsize=14, fontweight='bold', pad=12)
ax.set_xlabel('Class Label', fontsize=12)
ax.set_ylabel('Number of Samples', fontsize=12)
ax.set_ylim(0, max(raw_counts) * 1.15)
ax.grid(axis='y', alpha=0.3)
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)
plt.tight_layout()
save_fig('graph01_raw_class_distribution.png')


# ══════════════════════════════════════════════════════════════════════════════
# SECTION 3 — DATA AUGMENTATION & BALANCING  (Paper 1 §2.2 / Experiment 4)
# ══════════════════════════════════════════════════════════════════════════════
print(f"\n[2/9] Balancing classes to {TARGET_PER_CLASS} samples each …")

balanced_parts = []
for cls in CLASS_NAMES:
    cls_df = df[df['dx'] == cls].copy()
    n = len(cls_df)
    if n < TARGET_PER_CLASS:
        reps = (TARGET_PER_CLASS // n) + 1
        cls_df = pd.concat([cls_df] * reps, ignore_index=True).head(TARGET_PER_CLASS)
    else:
        cls_df = cls_df.head(TARGET_PER_CLASS)
    balanced_parts.append(cls_df)

balanced_df = (pd.concat(balanced_parts, ignore_index=True)
                 .sample(frac=1, random_state=RANDOM_SEED)
                 .reset_index(drop=True))

print(f"  Balanced total : {len(balanced_df):,}")
print("  Balanced distribution:")
for cls in CLASS_NAMES:
    n = (balanced_df['dx'] == cls).sum()
    print(f"    {cls:>6} : {n:>5}")

# ── Graph 2 — Balanced Class Distribution Bar Chart ──────────────────────────
bal_counts = [(balanced_df['dx'] == cls).sum() for cls in CLASS_NAMES]

fig, ax = plt.subplots(figsize=(9, 5))
bars = ax.bar(CLASS_NAMES, bal_counts, color=colors_bar, edgecolor='white', linewidth=0.8)
for bar, val in zip(bars, bal_counts):
    ax.text(bar.get_x() + bar.get_width()/2, bar.get_height() + 40,
            str(val), ha='center', va='bottom', fontsize=10, fontweight='bold')
ax.set_title('ResNet50 — Balanced Class Distribution', fontsize=14, fontweight='bold', pad=12)
ax.set_xlabel('Class Label', fontsize=12)
ax.set_ylabel('Number of Samples', fontsize=12)
ax.set_ylim(0, max(bal_counts) * 1.15)
ax.grid(axis='y', alpha=0.3)
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)
plt.tight_layout()
save_fig('graph02_balanced_class_distribution.png')

# ── Graph 3 — Before vs After Balancing (side-by-side) ───────────────────────
x   = np.arange(len(CLASS_NAMES))
w   = 0.35

fig, ax = plt.subplots(figsize=(11, 5))
b1 = ax.bar(x - w/2, raw_counts, w, label='Before Balancing', color='#4C72B0', alpha=0.85, edgecolor='white')
b2 = ax.bar(x + w/2, bal_counts, w, label='After Balancing',  color='#55A868', alpha=0.85, edgecolor='white')
ax.set_xticks(x); ax.set_xticklabels(CLASS_NAMES, fontsize=11)
ax.set_title('ResNet50 — Class Distribution Before vs After Balancing',
             fontsize=13, fontweight='bold', pad=12)
ax.set_xlabel('Class Label', fontsize=12)
ax.set_ylabel('Number of Samples', fontsize=12)
ax.legend(fontsize=11)
ax.grid(axis='y', alpha=0.3)
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)
plt.tight_layout()
save_fig('graph03_before_after_balancing.png')


# ══════════════════════════════════════════════════════════════════════════════
# SECTION 4 — TRAIN / VAL / TEST SPLIT  (80 / 10 / 10)
# ══════════════════════════════════════════════════════════════════════════════
print("\n[3/9] Splitting 80 / 10 / 10 (stratified) …")

train_df, temp_df = train_test_split(
    balanced_df, test_size=0.20,
    random_state=RANDOM_SEED, stratify=balanced_df['label']
)
val_df, test_df = train_test_split(
    temp_df, test_size=0.50,
    random_state=RANDOM_SEED, stratify=temp_df['label']
)

print(f"  Train : {len(train_df):>6,} samples")
print(f"  Val   : {len(val_df):>6,} samples")
print(f"  Test  : {len(test_df):>6,} samples")


# ══════════════════════════════════════════════════════════════════════════════
# SECTION 5 — DATA GENERATORS  (Paper 1 §2.2)
# FIX: ResNet50 requires Caffe-style preprocessing (RGB→BGR + ImageNet mean
# subtraction). Replaced rescale=1.0/255 with preprocessing_function=resnet_preprocess
# in both train and eval generators. All other augmentation parameters unchanged.
# ══════════════════════════════════════════════════════════════════════════════
print("\n[4/9] Building data generators …")

train_datagen = ImageDataGenerator(
    preprocessing_function=resnet_preprocess,
    rotation_range=20,
    width_shift_range=0.1,
    height_shift_range=0.1,
    zoom_range=0.1,
    horizontal_flip=True,
    vertical_flip=True,
    fill_mode='nearest'
)

eval_datagen = ImageDataGenerator(preprocessing_function=resnet_preprocess)

train_gen = train_datagen.flow_from_dataframe(
    train_df, x_col='path', y_col='dx',
    target_size=(IMG_SIZE, IMG_SIZE),
    batch_size=BATCH_SIZE,
    class_mode='categorical',
    classes=CLASS_NAMES,
    shuffle=True, seed=RANDOM_SEED
)

val_gen = eval_datagen.flow_from_dataframe(
    val_df, x_col='path', y_col='dx',
    target_size=(IMG_SIZE, IMG_SIZE),
    batch_size=BATCH_SIZE,
    class_mode='categorical',
    classes=CLASS_NAMES,
    shuffle=False
)

test_gen = eval_datagen.flow_from_dataframe(
    test_df, x_col='path', y_col='dx',
    target_size=(IMG_SIZE, IMG_SIZE),
    batch_size=BATCH_SIZE,
    class_mode='categorical',
    classes=CLASS_NAMES,
    shuffle=False
)

print(f"  Class index map : {train_gen.class_indices}")


# ══════════════════════════════════════════════════════════════════════════════
# SECTION 6 — BUILD ResNet50 MODEL  (Paper 1 §2.3, §2.4 / Figure 5)
# ══════════════════════════════════════════════════════════════════════════════
print("\n[5/9] Building ResNet50 model (Paper 1 Fig 5 architecture) …")

base = ResNet50(weights='imagenet', include_top=False,
                   input_shape=(IMG_SIZE, IMG_SIZE, 3))

for layer in base.layers:
    layer.trainable = False
for layer in base.layers[-30:]:
    layer.trainable = True

print(f"  Base layers (frozen)  : {sum(1 for l in base.layers if not l.trainable)}")
print(f"  Base layers (unfrozen): {sum(1 for l in base.layers if l.trainable)}")

x = base.output
x = GlobalAveragePooling2D(name='gap')(x)

x = Dense(512, name='dense_512')(x)
x = BatchNormalization(name='bn_512')(x)
x = Activation('relu', name='relu_512')(x)
x = Dropout(0.3, name='drop_512')(x)

x = Dense(128, name='dense_128')(x)
x = BatchNormalization(name='bn_128')(x)
x = Activation('relu', name='relu_128')(x)
x = Dropout(0.3, name='drop_128')(x)

x = Dense(64, name='dense_64')(x)
x = BatchNormalization(name='bn_64')(x)
x = Activation('relu', name='relu_64')(x)
x = Dropout(0.2, name='drop_64')(x)

x = Dense(32, name='dense_32')(x)
x = BatchNormalization(name='bn_32')(x)
x = Activation('relu', name='relu_32')(x)

output = Dense(NUM_CLASSES, activation='softmax', name='output')(x)

model = Model(inputs=base.input, outputs=output, name='ResNet50_Paper1')

print(f"  Total params      : {model.count_params():,}")
print(f"  Trainable params  : {sum(np.prod(v.shape) for v in model.trainable_variables):,}")

model.summary(line_length=80)


# ══════════════════════════════════════════════════════════════════════════════
# SECTION 7 — COMPILE
# ══════════════════════════════════════════════════════════════════════════════
model.compile(
    optimizer=Adamax(learning_rate=INITIAL_LR),
    loss='categorical_crossentropy',
    metrics=['accuracy']
)
print("\n[6/9] Model compiled  →  Adamax | lr=0.001 | categorical_crossentropy")


# ══════════════════════════════════════════════════════════════════════════════
# SECTION 8 — CALLBACKS
# ══════════════════════════════════════════════════════════════════════════════
print("\n[7/9] Configuring callbacks …")

def lr_schedule(epoch, lr):
    """Exponential decay after epoch 10 (Paper 1 §2.4)."""
    if epoch < 10:
        return float(lr)
    return float(lr * tf.math.exp(-0.05).numpy())

best_weights_path = os.path.join(OUT_DIR, 'resnet50_paper1_best.h5')

callbacks = [
    LearningRateScheduler(lr_schedule, verbose=0),
    ModelCheckpoint(
        best_weights_path,
        monitor='val_accuracy',
        save_best_only=True,
        verbose=1
    ),
    EarlyStopping(
        monitor='val_accuracy',
        patience=10,
        restore_best_weights=True,
        verbose=1
    )
]


# ══════════════════════════════════════════════════════════════════════════════
# SECTION 9 — TRAIN
# ══════════════════════════════════════════════════════════════════════════════
print("\n[8/9] Training …  (Experiment 4: balanced + tuned)")
print("─" * 65)

history = model.fit(
    train_gen,
    epochs=EPOCHS,
    validation_data=val_gen,
    callbacks=callbacks,
    verbose=1
)

if os.path.exists(best_weights_path):
    model.load_weights(best_weights_path)
    print("  ✓ Best weights restored.")


# ══════════════════════════════════════════════════════════════════════════════
# SECTION 10 — EVALUATION & ALL VISUALISATIONS
# ══════════════════════════════════════════════════════════════════════════════
print("\n[9/9] Evaluating on held-out test set …")
print("─" * 65)

test_loss, test_acc = model.evaluate(test_gen, verbose=0)
print(f"\n  ✅  Test Accuracy : {test_acc * 100:.2f}%")
print(f"  ✅  Test Loss     : {test_loss:.4f}")

test_gen.reset()
y_prob = model.predict(test_gen, verbose=1)
y_pred = np.argmax(y_prob, axis=1)
y_true = test_gen.classes

report     = classification_report(y_true, y_pred,
                                   target_names=CLASS_NAMES, output_dict=True)
report_df  = pd.DataFrame(report).T.round(4)

# ─────────────────────────────────────────────────────────────────────────────
# Graph 4 — Training Accuracy Curve
# ─────────────────────────────────────────────────────────────────────────────
fig, ax = plt.subplots(figsize=(9, 5))
ax.plot(history.history['accuracy'],     'b--', lw=2, label='Train Accuracy')
ax.plot(history.history['val_accuracy'], 'b-',  lw=2, label='Validation Accuracy')
ax.set_title('ResNet50 — Training Accuracy', fontsize=14, fontweight='bold', pad=12)
ax.set_xlabel('Epoch', fontsize=12)
ax.set_ylabel('Accuracy', fontsize=12)
ax.legend(fontsize=11)
ax.grid(True, alpha=0.3)
ax.set_ylim([0.70, 1.01])
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)
plt.tight_layout()
save_fig('graph04_training_accuracy.png')

# ─────────────────────────────────────────────────────────────────────────────
# Graph 5 — Training Loss Curve
# ─────────────────────────────────────────────────────────────────────────────
fig, ax = plt.subplots(figsize=(9, 5))
ax.plot(history.history['loss'],     'r--', lw=2, label='Train Loss')
ax.plot(history.history['val_loss'], 'r-',  lw=2, label='Validation Loss')
ax.set_title('ResNet50 — Training Loss', fontsize=14, fontweight='bold', pad=12)
ax.set_xlabel('Epoch', fontsize=12)
ax.set_ylabel('Loss', fontsize=12)
ax.legend(fontsize=11)
ax.grid(True, alpha=0.3)
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)
plt.tight_layout()
save_fig('graph05_training_loss.png')

# ─────────────────────────────────────────────────────────────────────────────
# Graph 6 — Accuracy & Loss Combined (Paper 1 Fig 6 style)
# ─────────────────────────────────────────────────────────────────────────────
fig, axes = plt.subplots(1, 2, figsize=(14, 5))
fig.suptitle('ResNet50 — Training & Validation Curves',
             fontsize=14, fontweight='bold')

axes[0].plot(history.history['accuracy'],     'b--', lw=2, label='ResNet50_Train')
axes[0].plot(history.history['val_accuracy'], 'b-',  lw=2, label='ResNet50_Validation')
axes[0].set_title('(a) Accuracy', fontsize=12)
axes[0].set_xlabel('Epochs'); axes[0].set_ylabel('Accuracy')
axes[0].legend(); axes[0].grid(True, alpha=0.3)
axes[0].set_ylim([0.70, 1.01])

axes[1].plot(history.history['loss'],     'r--', lw=2, label='ResNet50_Train')
axes[1].plot(history.history['val_loss'], 'r-',  lw=2, label='ResNet50_Validation')
axes[1].set_title('(b) Loss', fontsize=12)
axes[1].set_xlabel('Epochs'); axes[1].set_ylabel('Loss')
axes[1].legend(); axes[1].grid(True, alpha=0.3)

plt.tight_layout()
save_fig('graph06_training_curves_combined.png')

# ─────────────────────────────────────────────────────────────────────────────
# Graph 7 — Training Loss Over Epochs (standalone, larger)
# ─────────────────────────────────────────────────────────────────────────────
epochs_range = range(1, len(history.history['loss']) + 1)

fig, ax = plt.subplots(figsize=(10, 5))
ax.plot(epochs_range, history.history['loss'],     'r-o', lw=2,
        markersize=4, label='Train Loss')
ax.plot(epochs_range, history.history['val_loss'], 'r-s', lw=2,
        markersize=4, linestyle='--', label='Validation Loss')
ax.fill_between(epochs_range,
                history.history['loss'],
                history.history['val_loss'],
                alpha=0.08, color='red')
ax.set_title('ResNet50 — Training Loss Over Epochs',
             fontsize=14, fontweight='bold', pad=12)
ax.set_xlabel('Epoch', fontsize=12)
ax.set_ylabel('Categorical Cross-Entropy Loss', fontsize=12)
ax.legend(fontsize=11)
ax.grid(True, alpha=0.3)
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)
plt.tight_layout()
save_fig('graph07_loss_over_epochs_detailed.png')

# ─────────────────────────────────────────────────────────────────────────────
# Graph 8 — Normalised Confusion Matrix  (Paper 1 Fig 7)
# ─────────────────────────────────────────────────────────────────────────────
cm      = confusion_matrix(y_true, y_pred)
cm_norm = cm.astype(float) / cm.sum(axis=1, keepdims=True)

fig, ax = plt.subplots(figsize=(9, 7))
sns.heatmap(
    cm_norm, annot=True, fmt='.2f', cmap='Blues',
    xticklabels=CLASS_NAMES, yticklabels=CLASS_NAMES,
    linewidths=0.5, vmin=0, vmax=1, ax=ax
)
ax.set_title('ResNet50 — Confusion Matrix',
             fontsize=14, fontweight='bold', pad=12)
ax.set_ylabel('True Label',      fontsize=11)
ax.set_xlabel('Predicted Label', fontsize=11)
plt.tight_layout()
save_fig('graph08_confusion_matrix_normalised.png')

# ─────────────────────────────────────────────────────────────────────────────
# Graph 9 — Raw (Count) Confusion Matrix
# ─────────────────────────────────────────────────────────────────────────────
fig, ax = plt.subplots(figsize=(9, 7))
sns.heatmap(
    cm, annot=True, fmt='d', cmap='YlOrRd',
    xticklabels=CLASS_NAMES, yticklabels=CLASS_NAMES,
    linewidths=0.5, ax=ax
)
ax.set_title('ResNet50 — Confusion Matrix (Counts)',
             fontsize=14, fontweight='bold', pad=12)
ax.set_ylabel('True Label',      fontsize=11)
ax.set_xlabel('Predicted Label', fontsize=11)
plt.tight_layout()
save_fig('graph09_confusion_matrix_counts.png')

# ─────────────────────────────────────────────────────────────────────────────
# Graph 10 — ROC Curves per class  (Paper 1 Fig 8)
# ─────────────────────────────────────────────────────────────────────────────
y_true_bin = label_binarize(y_true, classes=list(range(NUM_CLASSES)))
COLORS     = ['blue', 'red', 'green', 'darkorange', 'purple', 'brown', 'deeppink']

fig, ax = plt.subplots(figsize=(10, 8))
all_fpr   = np.unique(np.concatenate([roc_curve(y_true_bin[:, i], y_prob[:, i])[0]
                                       for i in range(NUM_CLASSES)]))
mean_tpr  = np.zeros_like(all_fpr)

roc_aucs = []
for i, (cls, col) in enumerate(zip(CLASS_NAMES, COLORS)):
    fpr, tpr, _ = roc_curve(y_true_bin[:, i], y_prob[:, i])
    roc_auc     = auc(fpr, tpr)
    roc_aucs.append(roc_auc)
    n_pos = y_true_bin[:, i].sum()
    n_neg = len(y_true_bin) - n_pos
    se    = np.sqrt(roc_auc * (1 - roc_auc) *
                    (1/n_pos + 1/n_neg)) if (n_pos > 0 and n_neg > 0) else 0
    ci_lo = max(0.0, roc_auc - 1.96 * se)
    ci_hi = min(1.0, roc_auc + 1.96 * se)
    ax.plot(fpr, tpr, color=col, lw=2,
            label=f'Class {i} ({cls}): AUC={roc_auc:.3f} (95% CI: {ci_lo:.3f}–{ci_hi:.3f})')
    mean_tpr += np.interp(all_fpr, fpr, tpr)

mean_tpr  /= NUM_CLASSES
mean_auc   = auc(all_fpr, mean_tpr)
ax.plot(all_fpr, mean_tpr, 'k-', lw=2.5, label=f'Macro-avg AUC = {mean_auc:.3f}')
ax.plot([0, 1], [0, 1], 'k--', lw=1)
ax.set_xlim([0, 1]); ax.set_ylim([0, 1.02])
ax.set_xlabel('False Positive Rate', fontsize=12)
ax.set_ylabel('True Positive Rate',  fontsize=12)
ax.set_title('ResNet50 — ROC Curves',
             fontsize=14, fontweight='bold', pad=12)
ax.legend(loc='lower right', fontsize=8)
ax.grid(True, alpha=0.3)
plt.tight_layout()
save_fig('graph10_roc_curves.png')

# ─────────────────────────────────────────────────────────────────────────────
# Graph 11 — Per-Class AUC Bar Chart
# ─────────────────────────────────────────────────────────────────────────────
fig, ax = plt.subplots(figsize=(9, 5))
bars = ax.bar(CLASS_NAMES, roc_aucs, color=COLORS, edgecolor='white', linewidth=0.8)
for bar, val in zip(bars, roc_aucs):
    ax.text(bar.get_x() + bar.get_width()/2, bar.get_height() + 0.003,
            f'{val:.3f}', ha='center', va='bottom', fontsize=10, fontweight='bold')
ax.axhline(y=mean_auc, color='black', linestyle='--', lw=1.5,
           label=f'Macro Avg AUC = {mean_auc:.3f}')
ax.set_title('ResNet50 — Per-Class AUC',
             fontsize=14, fontweight='bold', pad=12)
ax.set_xlabel('Class Label', fontsize=12)
ax.set_ylabel('AUC Score', fontsize=12)
ax.set_ylim([0.85, 1.02])
ax.legend(fontsize=11)
ax.grid(axis='y', alpha=0.3)
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)
plt.tight_layout()
save_fig('graph11_per_class_auc.png')

# ─────────────────────────────────────────────────────────────────────────────
# Graph 12 — Per-Class Precision Bar Chart
# ─────────────────────────────────────────────────────────────────────────────
per_class_precision = [report[cls]['precision'] for cls in CLASS_NAMES]
macro_prec          = report['macro avg']['precision']

fig, ax = plt.subplots(figsize=(9, 5))
bars = ax.bar(CLASS_NAMES, per_class_precision, color=colors_bar, edgecolor='white')
for bar, val in zip(bars, per_class_precision):
    ax.text(bar.get_x() + bar.get_width()/2, bar.get_height() + 0.005,
            f'{val:.3f}', ha='center', va='bottom', fontsize=10, fontweight='bold')
ax.axhline(y=macro_prec, color='black', linestyle='--', lw=1.5,
           label=f'Macro Avg = {macro_prec:.3f}')
ax.set_title('ResNet50 — Per-Class Precision',
             fontsize=14, fontweight='bold', pad=12)
ax.set_xlabel('Class Label', fontsize=12)
ax.set_ylabel('Precision', fontsize=12)
ax.set_ylim([0, 1.10])
ax.legend(fontsize=11)
ax.grid(axis='y', alpha=0.3)
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)
plt.tight_layout()
save_fig('graph12_per_class_precision.png')

# ─────────────────────────────────────────────────────────────────────────────
# Graph 13 — Per-Class Recall Bar Chart
# ─────────────────────────────────────────────────────────────────────────────
per_class_recall = [report[cls]['recall'] for cls in CLASS_NAMES]
macro_rec        = report['macro avg']['recall']

fig, ax = plt.subplots(figsize=(9, 5))
bars = ax.bar(CLASS_NAMES, per_class_recall, color=colors_bar, edgecolor='white')
for bar, val in zip(bars, per_class_recall):
    ax.text(bar.get_x() + bar.get_width()/2, bar.get_height() + 0.005,
            f'{val:.3f}', ha='center', va='bottom', fontsize=10, fontweight='bold')
ax.axhline(y=macro_rec, color='black', linestyle='--', lw=1.5,
           label=f'Macro Avg = {macro_rec:.3f}')
ax.set_title('ResNet50 — Per-Class Recall',
             fontsize=14, fontweight='bold', pad=12)
ax.set_xlabel('Class Label', fontsize=12)
ax.set_ylabel('Recall (Sensitivity)', fontsize=12)
ax.set_ylim([0, 1.10])
ax.legend(fontsize=11)
ax.grid(axis='y', alpha=0.3)
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)
plt.tight_layout()
save_fig('graph13_per_class_recall.png')

# ─────────────────────────────────────────────────────────────────────────────
# Graph 14 — Per-Class F1-Score Bar Chart
# ─────────────────────────────────────────────────────────────────────────────
per_class_f1 = [report[cls]['f1-score'] for cls in CLASS_NAMES]
macro_f1     = report['macro avg']['f1-score']

fig, ax = plt.subplots(figsize=(9, 5))
bars = ax.bar(CLASS_NAMES, per_class_f1, color=colors_bar, edgecolor='white')
for bar, val in zip(bars, per_class_f1):
    ax.text(bar.get_x() + bar.get_width()/2, bar.get_height() + 0.005,
            f'{val:.3f}', ha='center', va='bottom', fontsize=10, fontweight='bold')
ax.axhline(y=macro_f1, color='black', linestyle='--', lw=1.5,
           label=f'Macro Avg = {macro_f1:.3f}')
ax.set_title('ResNet50 — Per-Class F1-Score',
             fontsize=14, fontweight='bold', pad=12)
ax.set_xlabel('Class Label', fontsize=12)
ax.set_ylabel('F1-Score', fontsize=12)
ax.set_ylim([0, 1.10])
ax.legend(fontsize=11)
ax.grid(axis='y', alpha=0.3)
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)
plt.tight_layout()
save_fig('graph14_per_class_f1.png')

# ─────────────────────────────────────────────────────────────────────────────
# Graph 15 — Grouped Bar Chart: Precision, Recall, F1 per Class
# ─────────────────────────────────────────────────────────────────────────────
x = np.arange(len(CLASS_NAMES))
w = 0.25

fig, ax = plt.subplots(figsize=(13, 6))
b1 = ax.bar(x - w,     per_class_precision, w, label='Precision', color='#4C72B0', edgecolor='white')
b2 = ax.bar(x,         per_class_recall,    w, label='Recall',    color='#55A868', edgecolor='white')
b3 = ax.bar(x + w,     per_class_f1,        w, label='F1-Score',  color='#C44E52', edgecolor='white')

for bars in [b1, b2, b3]:
    for bar in bars:
        h = bar.get_height()
        ax.text(bar.get_x() + bar.get_width()/2, h + 0.005,
                f'{h:.2f}', ha='center', va='bottom', fontsize=7.5, fontweight='bold')

ax.set_xticks(x); ax.set_xticklabels(CLASS_NAMES, fontsize=11)
ax.set_title('ResNet50 — Per-Class Precision, Recall & F1-Score',
             fontsize=13, fontweight='bold', pad=12)
ax.set_xlabel('Class Label', fontsize=12)
ax.set_ylabel('Score', fontsize=12)
ax.set_ylim([0, 1.15])
ax.legend(fontsize=11)
ax.grid(axis='y', alpha=0.3)
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)
plt.tight_layout()
save_fig('graph15_grouped_precision_recall_f1.png')

# ─────────────────────────────────────────────────────────────────────────────
# Graph 16 — Macro-Average Metrics Summary Bar Chart
# ─────────────────────────────────────────────────────────────────────────────
macro_metrics  = {
    'Accuracy'  : test_acc,
    'Precision' : macro_prec,
    'Recall'    : macro_rec,
    'F1-Score'  : macro_f1,
    'Macro AUC' : mean_auc,
}
metric_colors = ['#4C72B0', '#DD8452', '#55A868', '#C44E52', '#8172B2']

fig, ax = plt.subplots(figsize=(9, 5))
bars = ax.bar(macro_metrics.keys(), macro_metrics.values(),
              color=metric_colors, edgecolor='white', linewidth=0.8, width=0.5)
for bar, (k, v) in zip(bars, macro_metrics.items()):
    ax.text(bar.get_x() + bar.get_width()/2, bar.get_height() + 0.003,
            f'{v:.4f}', ha='center', va='bottom', fontsize=11, fontweight='bold')
ax.set_title('ResNet50 — Overall Performance Metrics',
             fontsize=14, fontweight='bold', pad=12)
ax.set_ylabel('Score', fontsize=12)
ax.set_ylim([0, 1.12])
ax.grid(axis='y', alpha=0.3)
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)
plt.tight_layout()
save_fig('graph16_overall_metrics_summary.png')

# ─────────────────────────────────────────────────────────────────────────────
# Graph 17 — Per-Class Accuracy (Diagonal of normalised CM)
# ─────────────────────────────────────────────────────────────────────────────
per_class_acc = np.diag(cm_norm)

fig, ax = plt.subplots(figsize=(9, 5))
bars = ax.bar(CLASS_NAMES, per_class_acc, color=colors_bar, edgecolor='white')
for bar, val in zip(bars, per_class_acc):
    ax.text(bar.get_x() + bar.get_width()/2, bar.get_height() + 0.005,
            f'{val:.3f}', ha='center', va='bottom', fontsize=10, fontweight='bold')
ax.axhline(y=np.mean(per_class_acc), color='black', linestyle='--', lw=1.5,
           label=f'Mean = {np.mean(per_class_acc):.3f}')
ax.set_title('ResNet50 — Per-Class Accuracy',
             fontsize=14, fontweight='bold', pad=12)
ax.set_xlabel('Class Label', fontsize=12)
ax.set_ylabel('Accuracy (True Positive Rate)', fontsize=12)
ax.set_ylim([0, 1.15])
ax.legend(fontsize=11)
ax.grid(axis='y', alpha=0.3)
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)
plt.tight_layout()
save_fig('graph17_per_class_accuracy.png')

# ─────────────────────────────────────────────────────────────────────────────
# Graph 18 — Learning Rate Schedule
# ─────────────────────────────────────────────────────────────────────────────
if 'lr' in history.history:
    fig, ax = plt.subplots(figsize=(9, 4))
    ax.plot(history.history['lr'], 'g-o', markersize=4, lw=2)
    ax.set_title('ResNet50 — Learning Rate Schedule',
                 fontsize=14, fontweight='bold', pad=12)
    ax.set_xlabel('Epoch', fontsize=12)
    ax.set_ylabel('Learning Rate', fontsize=12)
    ax.grid(True, alpha=0.3)
    ax.spines['top'].set_visible(False)
    ax.spines['right'].set_visible(False)
    plt.tight_layout()
    save_fig('graph18_lr_schedule.png')

# ─────────────────────────────────────────────────────────────────────────────
# Graph 19 — Accuracy Gap (Train vs Val) — Overfitting Monitor
# ─────────────────────────────────────────────────────────────────────────────
train_acc_hist = history.history['accuracy']
val_acc_hist   = history.history['val_accuracy']
gap            = [t - v for t, v in zip(train_acc_hist, val_acc_hist)]

fig, ax = plt.subplots(figsize=(9, 4))
ax.bar(range(1, len(gap)+1), gap,
       color=['#C44E52' if g > 0.05 else '#55A868' for g in gap],
       edgecolor='white')
ax.axhline(y=0.05, color='red', linestyle='--', lw=1.5, label='5% threshold')
ax.set_title('ResNet50 — Train–Validation Accuracy Gap (Overfitting Monitor)',
             fontsize=13, fontweight='bold', pad=12)
ax.set_xlabel('Epoch', fontsize=12)
ax.set_ylabel('Train Acc − Val Acc', fontsize=12)
ax.legend(fontsize=11)
ax.grid(axis='y', alpha=0.3)
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)
plt.tight_layout()
save_fig('graph19_overfitting_monitor.png')

# ─────────────────────────────────────────────────────────────────────────────
# Graph 20 — Prediction Confidence Distribution (all classes)
# ─────────────────────────────────────────────────────────────────────────────
max_probs = np.max(y_prob, axis=1)
correct   = (y_pred == y_true)

fig, ax = plt.subplots(figsize=(9, 5))
ax.hist(max_probs[correct],  bins=30, alpha=0.7, color='#55A868', label='Correct')
ax.hist(max_probs[~correct], bins=30, alpha=0.7, color='#C44E52', label='Incorrect')
ax.set_title('ResNet50 — Prediction Confidence Distribution',
             fontsize=14, fontweight='bold', pad=12)
ax.set_xlabel('Max Softmax Probability (Confidence)', fontsize=12)
ax.set_ylabel('Number of Samples', fontsize=12)
ax.legend(fontsize=11)
ax.grid(True, alpha=0.3)
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)
plt.tight_layout()
save_fig('graph20_confidence_distribution.png')

# ─────────────────────────────────────────────────────────────────────────────
# Graph 21 — Per-Class Prediction Count (Predicted vs True)
# ─────────────────────────────────────────────────────────────────────────────
true_counts      = [np.sum(y_true == i) for i in range(NUM_CLASSES)]
predicted_counts = [np.sum(y_pred == i) for i in range(NUM_CLASSES)]
x = np.arange(len(CLASS_NAMES)); w = 0.35

fig, ax = plt.subplots(figsize=(10, 5))
b1 = ax.bar(x - w/2, true_counts,      w, label='Actual',    color='#4C72B0', edgecolor='white')
b2 = ax.bar(x + w/2, predicted_counts, w, label='Predicted', color='#DD8452', edgecolor='white')
ax.set_xticks(x); ax.set_xticklabels(CLASS_NAMES, fontsize=11)
ax.set_title('ResNet50 — Actual vs Predicted Class Counts',
             fontsize=13, fontweight='bold', pad=12)
ax.set_xlabel('Class Label', fontsize=12)
ax.set_ylabel('Number of Samples', fontsize=12)
ax.legend(fontsize=11)
ax.grid(axis='y', alpha=0.3)
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)
plt.tight_layout()
save_fig('graph21_actual_vs_predicted_counts.png')

# ─────────────────────────────────────────────────────────────────────────────
# Graph 22 — Macro-Average vs Weighted-Average Metrics Comparison
# ─────────────────────────────────────────────────────────────────────────────
weighted_f1   = report['weighted avg']['f1-score']
weighted_prec = report['weighted avg']['precision']
weighted_rec  = report['weighted avg']['recall']

categories = ['Precision', 'Recall', 'F1-Score']
macro_vals  = [macro_prec, macro_rec, macro_f1]
weighted_vals = [weighted_prec, weighted_rec, weighted_f1]

x = np.arange(len(categories)); w = 0.30
fig, ax = plt.subplots(figsize=(8, 5))
b1 = ax.bar(x - w/2, macro_vals,    w, label='Macro Average',    color='#4C72B0', edgecolor='white')
b2 = ax.bar(x + w/2, weighted_vals, w, label='Weighted Average', color='#DD8452', edgecolor='white')
for bars in [b1, b2]:
    for bar in bars:
        h = bar.get_height()
        ax.text(bar.get_x() + bar.get_width()/2, h + 0.004,
                f'{h:.4f}', ha='center', va='bottom', fontsize=9, fontweight='bold')
ax.set_xticks(x); ax.set_xticklabels(categories, fontsize=12)
ax.set_title('ResNet50 — Macro vs Weighted Average Metrics',
             fontsize=13, fontweight='bold', pad=12)
ax.set_ylabel('Score', fontsize=12)
ax.set_ylim([0, 1.12])
ax.legend(fontsize=11)
ax.grid(axis='y', alpha=0.3)
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)
plt.tight_layout()
save_fig('graph22_macro_vs_weighted_metrics.png')

# ─────────────────────────────────────────────────────────────────────────────
# Save metrics CSV
# ─────────────────────────────────────────────────────────────────────────────
report_df.to_csv(os.path.join(FIG_DIR, 'table_metrics.csv'))
print("\n  Saved → table_metrics.csv")

# ── Classification report ─────────────────────────────────────────────────────
print("\n" + "=" * 65)
print("  CLASSIFICATION REPORT")
print("=" * 65)
print(classification_report(y_true, y_pred, target_names=CLASS_NAMES, digits=4))


# ══════════════════════════════════════════════════════════════════════════════
# SAVE MODEL
# ══════════════════════════════════════════════════════════════════════════════
model.save(os.path.join(OUT_DIR, 'resnet50_paper1_final.h5'))
print("\n  Model saved → resnet50_paper1_final.h5")


# ══════════════════════════════════════════════════════════════════════════════
# PACKAGE ALL FIGURES INTO A ZIP FILE
# ══════════════════════════════════════════════════════════════════════════════
ZIP_PATH = os.path.join(OUT_DIR, 'resnet50_thesis_figures.zip')

with zipfile.ZipFile(ZIP_PATH, 'w', zipfile.ZIP_DEFLATED) as zf:
    for fname in sorted(os.listdir(FIG_DIR)):
        fpath = os.path.join(FIG_DIR, fname)
        if os.path.isfile(fpath):
            zf.write(fpath, arcname=fname)
            print(f"  Zipped → {fname}")

print(f"\n  ✅  All figures saved to : {ZIP_PATH}")
print(f"  Download via Kaggle Output panel: resnet50_thesis_figures.zip")


# ══════════════════════════════════════════════════════════════════════════════
# FINAL RESULTS SUMMARY
# ══════════════════════════════════════════════════════════════════════════════
print("\n" + "=" * 65)
print("  FINAL RESULTS SUMMARY")
print("=" * 65)
print(f"  Paper        : Villa-Pulgarin et al. CMC 2022 (Experiment 4)")
print(f"  Model        : ResNet50 (Optimized)")
print(f"  Dataset      : HAM10000  |  7 classes  |  Split: 80/10/10")
print(f"  Input Size   : {IMG_SIZE}×{IMG_SIZE}×3")
print(f"  Optimizer    : Adamax  |  Initial LR: {INITIAL_LR}")
print(f"  Epochs       : {EPOCHS}  |  Batch Size: {BATCH_SIZE}")
print(f"  Architecture : ResNet50 → GAP → Dense[512→128→64→32] → Softmax(7)")
print(f"  BN + ReLU between every dense block  |  Dropout: 0.3/0.3/0.2")
print("─" * 65)
print(f"  Test Accuracy     : {test_acc  * 100:.2f} %")
print(f"  Test Loss         : {test_loss:.4f}")
print(f"  Macro Avg F1      : {macro_f1:.4f}")
print(f"  Weighted Avg F1   : {weighted_f1:.4f}")
print(f"  Macro Avg Prec    : {macro_prec:.4f}")
print(f"  Macro Avg Recall  : {macro_rec:.4f}")
print(f"  Macro AUC (ROC)   : {mean_auc:.4f}")
print("─" * 65)
print(f"  Paper 1 reported  : ResNet50 Optimized = 96 %  (Table 7)")
print("=" * 65)
print(f"\n  Figures zipped  → /kaggle/working/resnet50_thesis_figures.zip")
print(f"  Graphs included : 22 thesis-ready figures")
```
